```python
# **문제 6. 불린 인덱싱의 괄호**
# 'A' 컬럼이 50보다 크고 'B' 컬럼이 100보다 작은 데이터를 필터링하려는데 에러가 납니다.
import pandas as pd

data = {
    'row1': [10, 20, 60],
    'row2': [30, 40, 130]
}
df = pd.DataFrame(data)

df

# 에러 발생 코드
filtered = df[df['row1'] > 50 & df['row2'] < 100] #불린 인덱싱은 괄호()를 붙여주어야 함
filtered
```

```python
import pandas as pd
import numpy as np

# **문제 9. 특정 조건 데이터 수정**
# `df`에서 '등급(grade)'이 'A'인 행들의 '보너스(bonus)' 컬럼 값을 일괄적으로 500으로 변경하는 코드를 작성하세요. (힌트: `loc` 사용)

# df를 100개의 행과 랜덤 grade, bonus를 가지도록 수정
df = pd.DataFrame({
    'grade' : np.random.choice(['A', 'B', 'C'], size=100),
    'bonus' : np.random.randint(50, 300, size=100)
})

df.loc[df['grade'] == 'A', 'bonus'] = 500 #loc는 할당
df

#loc 조건사용법
# 단일 조건: df.loc[df['Age'] > 24] (Age가 24 초과인 행 전체).
# 다중 조건: df.loc[(df['Age'] > 24) & (df['Type'] == 'A')] (괄호 필수).
# 행/열 동시 필터링: df.loc[df['Age'] > 24, ['Name', 'Age']].
# 조건부 값 변경: df.loc[df['Age'] < 20, 'Status'] = 'Junior'.
```
