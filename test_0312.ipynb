```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [23, 27, 31, 29],
    "Score": [75, 82, 90, 68]
})

df["Pass"] = np.where(df["Score"] >= 80, "Pass", "Fail")

df.head()
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [23, 27, 31, 29],
    "Score": [75, 82, 90, 68]
})

df["Level"] = np.where(df["Score"] >= df["Score"].mean(), "High", "Low")
df.head()
```

```python
#10번문제 풀이 필요

import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [23, 27, 31, 29],
    "Score": [75, 82, 90, 68]
})

import matplotlib.pyplot as plt

# 히스토그램 그리기
plt.figure(figsize=(8, 5))
plt.hist(df['Score'], bins=5, color='skyblue', edgecolor='black')

# 그래프 제목 및 축 라벨 설정
plt.title('Distribution of Scores')
plt.xlabel('Score')
plt.ylabel('Frequency')

# 그래프 표시
plt.grid(axis='y', alpha=0.3)
plt.show()
```
