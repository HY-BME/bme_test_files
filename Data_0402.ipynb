```python
import pandas as pd

df = pd.read_csv('/content/sample_data/avocado.csv')

df.head()
# df.shape
# df.columns
# df.info()
# df.describe()

# df['region'].value_counts()

#agg : 여러 통계를 한번에 구하는 메소드
region_year_summary = df.groupby(['region', 'year']).agg({
    'AveragePrice':'mean',
    'Total Volume':'sum'
})

#내림차순 정렬
# print(region_year_summary.sort_values('Total Volume', ascending=False).head(10))
print(region_year_summary.head(10))
```

```python
# 3. 지역, 연도, 유기농 여부 세 가지로 그룹화
final_summary = df.groupby(['region', 'year', 'type']).agg({
    'AveragePrice': 'mean',
    'Total Volume': 'sum'
})

# 텍스트로 보기 편하게 '행'을 '열'로 보낼 수도 있습니다 (unstack 활용)
# 예: 유기농 여부를 컬럼으로 빼서 비교하기
comparison = final_summary['AveragePrice'].unstack()
print(comparison.head(10))
```

```python
#Date 날짜 형변환

# 1. Date 컬럼을 datetime 타입으로 변환
df['Date'] = pd.to_datetime(df['Date'])
df['year_extracted'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df['day'] = df['Date'].dt.day

# 요일 정보 추출 (0: 월요일, 6: 일요일)
df['day_of_week'] = df['Date'].dt.day_name()
print(df[['Date', 'year_extracted', 'month', 'day', 'day_of_week']].head(),"\n\n")

#정렬
df = df.sort_values(by='Date')

# 2017년 하반기(7월 이후) 데이터만 추출
df_second_half_2017 = df[df['Date'] >= '2017-07-01']

# 특정 년,월(예: 2015년 2월) 데이터만 보기
february_data = df[(df['Date'].dt.year == 2015) & (df['Date'].dt.month == 2)]
print(february_data.head)
```

```python
import pandas as pd

df = pd.read_csv('/content/sample_data/Healthcare.csv')

df.info()

# 나이 내림차순 정렬 (큰 숫자부터)
age_sorted = df.sort_values(by='Age', ascending=False)

# 질병 오름차순, 증상 개수는 내림차순 정렬
complex_sorted = df.sort_values(by=['Disease', 'Symptom_Count'], ascending=[True, False])

#fever, cough, appetite loss, dizziness, diarrh. - 비정형 데이터
# 질병별 평균 계산
disease_stats = df.groupby('Disease')[['Age', 'Symptom_Count']].mean()

# 성별(Gender)에 따른 나이의 평균과 증상 개수의 최대값 구하기
gender_analysis = df.groupby('Gender').agg({
    'Age': 'mean',
    'Symptom_Count': 'max',
    'Patient_ID': 'count'  # 그룹별 환자 수
})
print(gender_analysis)
```

```python
result = (df
    .groupby('Disease')[['Age']]      # 1. 질병별로 나이 선택
    .mean()                           # 2. 평균 계산
    .sort_values(by='Age', ascending=False)  # 3. 평균 나이 높은 순 정렬
    .head(5)                          # 4. 상위 5개 질병 확인
)

# print("--- 평균 연령이 높은 상위 5개 질병 ---")
# print(result)

Patients_top5 = (df
    .groupby('Disease')['Patient_ID']      # 1. 질병별로 환자 ID 선택
    .count()                               # 2. 각 질병별 환자 수 계산
    .reset_index(name='Patient_Count')     # 3. Series를 DataFrame으로 변환하고 컬럼 이름 지정
    .sort_values(by='Patient_Count', ascending=False)  # 4. 환자 수 내림차순 정렬
    .head(5)                               # 5. 상위 5개 질병 확인
)

print("--- 환자가 많은 상위 5개 질병 ---")
print(Patients_top5)
```
