```python
# 기대 수명이 70세 이상이면 'High', 아니면 'Low'로 분류하는 함수
def categorize_life(exp):
    if exp >= 35:
        return 'High'
    else:
        return 'Low'

#apply 새로운 컬럼 생성
df['life_level'] = df['lifeExp'].apply(categorize_life)
df[['country', 'lifeExp', 'life_level']].head(10)
```

```python
# 1.  "대륙별로 가장 인구가 많은 나라는 어디인지" 찾아보자.
continent_pop = df.groupby('continent')['pop'].max()

def max_pop_country(continent):
    max_pop_data = df[(df['continent'] == continent) & (df['pop'] == continent_pop[continent])]
    return max_pop_data['country'].values[0]

print(max_pop_country('Asia'))
```

```python
# 1. 메서드 체이닝 (Method Chaining)

# 괄호()로 전체를 감싸면 줄바꿈을 자유롭게 할 수 있어 가독성이 좋아집니다.
#query - filter 순서
result = (df
    .query('country == "Korea, Rep."')  # 행 필터링
    .filter(items=['year', 'lifeExp'])  # 열 선택
    .head(3)                            # 상위 3개 확인
)

test = (df
          .query('country == "Croatia"')
          .filter(items=['pop', 'gdpPercap'])
          .head(3))
print(test)
```

```python
# #### `.pipe()` 메서드 (진정한 파이프 연산)
# 단순한 판다스 내장 함수가 아니라, **내가 만든 커스텀 함수**를 연속적으로 적용하고 싶을 때 사용합니다. R의 `%>%`와 가장 흡사한 구조를 가집니다.

# 1. 데이터 정제 함수 정의
def extract_asia(data):
    return data[data['continent'] == 'Asia']

# 2. 추가 계산 함수 정의
def add_gdp_k(data):
    data['gdp_k'] = data['gdpPercap'] / 1000
    return data

# 3. .pipe()로 연결하기
processed_df = (df
    .pipe(extract_asia)  # 아시아 데이터만 추출
    .pipe(add_gdp_k)     # GDP 단위 변경 열 추가
    .query('year == 2007') # 2007년 데이터만 선택
)

print(processed_df.head())
```

```python
def extract_asia(data):
    # .copy()를 사용하여 명시적으로 새로운 복사본을 생성합니다.
    return data[data['continent'] == 'Asia'].copy()

def add_gdp_k(data):
    # 이제 data는 독립된 복사본이므로 안전하게 열을 추가할 수 있습니다.
    data['gdp_k'] = data['gdpPercap'] / 1000
    return data

# 다시 실행
processed_df = (
    df
    .pipe(extract_asia)
    .pipe(add_gdp_k)
    .query('year == 2007')
)

processed_df.head()
```

```python
#특정 대륙별로 평균 수명을 구하고 소수점 둘째 자리까지 반올림

# 1. 대륙별 그룹화 및 평균 계산 함수
def get_continent_life_mean(data):
    return data.groupby(['continent'])['lifeExp'].mean()

# 2. 반올림 처리 함수
def round_series(data, decimals=2): #소수점 둘째 자리
    return data.round(decimals)

def year_condition(data):
    return data.loc[2007] #index로 년도를 찾음

# 3. 파이프라인 연결
continent_life_rounded = (
    df
    .query('year == 2007')
    .pipe(get_continent_life_mean)
    .pipe(round_series, decimals=2)
    #.pipe(year_condition)
)

continent_life_rounded
```
