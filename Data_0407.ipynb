```python
import pandas as pd

df = pd.read_csv('/content/sample_data/Healthcare.csv')

# 1. 'fever'라는 단어가 포함된 행만 필터링
fever_patients = df[df['Symptoms'].str.contains('fever', case=False, na=False)]

# 2. 결과 확인 (fever 환자들은 어떤 질병에 많이 걸렸나?)
fever_patients['Disease'].value_counts().head() #value_counts 자동정렬
```

```python
# [정제 및 추출 파이프라인]
cleaned_fever_analysis = (df
    # A. 텍스트 정제: 소문자로 바꾸고(lower) 양끝 공백 제거(strip)
    .assign(Symptoms = lambda x: x['Symptoms'].str.lower().str.strip())

    # B. 특정 증상 추출: 'fever' 또는 'cough'가 포함된 경우
    .query("Symptoms.str.contains('fever | cough')")

    # C. 분석: 질병별 평균 증상 개수 확인
    .groupby('Disease')['Symptom_Count'].mean()
    .sort_values(ascending=False)
)

print(cleaned_fever_analysis.head())
```

```python
# 1. 쉼표를 기준으로 문자열을 리스트로 분리
df['Symptom_List'] = df['Symptoms'].str.split(',')

# 2. 분리된 리스트의 첫 번째 증상만 따로 뽑기
df['Main_Symptom'] = df['Symptom_List'].str[0]

print(df[['Symptoms', 'Main_Symptom']].head())
```

```python
# 만약 증상명이 너무 복잡하거나(예: "mild fever with chills"), 비슷한 의미의 단어를 한꺼번에 정제하고 싶다면 아래와 같은 방식을 사용하세요.

#함수 사용
def categorize_symptom(text):
  text = text.lower()
  if 'fever' in text or 'sweating' in text:
    return 'Heat-related'
  elif  'cough' in text or 'sore throat' in text:
    return 'Respiratory'
  else:
     return 'Other'

df['Symptom_Type'] = df['Symptoms'].apply(categorize_symptom)
print(df['Symptom_Type'].value_counts())
```

```python
import pandas as pd

# Gapminder 데이터셋 URL (탭으로 구분된 tsv 파일인 경우 sep='\t' 추가)
url = "https://raw.githubusercontent.com/jennybc/gapminder/master/inst/extdata/gapminder.tsv"
df = pd.read_csv(url, sep='\t')

# 데이터의 앞부분 5줄만 확인
print(df.head())

# 무엇을 기록했는지는 컬럼으로 구분. 한 행은 하나의 샘플 데이터 - DataFrame의 일반적 관행
# wideFrame Data 형식 : 한 행이 하나의 기록치가 아닌 한 열이 기록치로 분석하기 어려움 -> 병합.재구성

# Wide 형태의 데이터를 Long 형태로 변환하는 가장 대표적인 방법은 pd.melt() 함수를 사용하는 것입니다.
# 이 함수는 식별자 변수(id_vars)를 고정하고, 나머지 컬럼들을 '변수(variable)'와 '값(value)'이라는 두 개의 열로 녹여내어(melt) 세로로 긴 구조를 만듭니다.
```

```python
df_gen = pd.read_csv('/content/sample_data/electricity_generation_per_person.csv')
df_use = pd.read_csv('/content/sample_data/electricity_use_per_person.csv')

df_gen = df_gen.drop(columns=['geo'])
df_use = df_use.drop(columns=['geo'])

df_use.head()
```

```python
import pandas as pd

# df_gen 변환: 'name'을 고정하고 나머지 연도 컬럼을 행으로 녹임
df_gen_long = pd.melt(df_gen,
                       id_vars=['name'],
                       var_name='year',
                       value_name='generation')

# 'name' 컬럼명을 'country'로 변경
df_gen_long = df_gen_long.rename(columns={'name': 'country'})

# df_use 변환: value_name을 'use'로 지정
df_use_long = pd.melt(df_use,
                      id_vars=['name'],
                      var_name='year',
                      value_name='use')

# df_use_long도 일관성을 위해 변경
df_use_long = df_use_long.rename(columns={'name': 'country'})

display(df_gen_long.head())
display(df_use_long.head())

print(df_gen_long.info())
print(df_use_long.info())
```

```python
# 수정된 컬럼명(generation, use)을 바탕으로 병합
df_combined = pd.merge(df_gen_long, df_use_long, on=['country', 'year'], how='inner')

# 결과 확인
display(df_combined.head(20))
print(df_combined.info())
```
