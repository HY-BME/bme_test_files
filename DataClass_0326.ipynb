```python
def clean_and_log(df):
    old_count = len(df) #행의 개수 = 데이터의 개수

    # 예: 가격이 0원인 이상한 데이터 삭제
    df = df[df['가격'] > 0]

    new_count = len(df)
    print(f"[정제 완료] 삭제된 데이터: {old_count - new_count}건")
    return df

df = pd.DataFrame({
    '상품': ['A', 'B', 'C'],
     '가격': [100, np.nan, 150]
    })

clean_and_log(df)
```

```python
import re

sentence = "제 전화번호는 010-1234-5678 입니다."
sentence_pattern = r"\d{3}-\d{4}-\d{4}"

# AI가 짜준 정규표현식 코드
def hide_phone(text):
    return re.sub(sentence_pattern, '[삭제]', text) #sub

print(hide_phone(sentence))
```

```python
import pandas as pd

# 테스트용 데이터프레임 생성
data = {
    '주문번호': [101, 102, 103, 104],
    '상품': ['노트북', '마우스', '키보드', '모니터'],
    '코드': [0, 1, 0, 1]  # 0: 미결제, 1: 결제완료
}
df = pd.DataFrame(data)

# 기존 코드 실행: map을 이용한 상태 컬럼 생성
df['상태'] = df['코드'].map({0: '미결제', 1: '결제완료'})

# 결과 확인
display(df)
```

```python
#공공데이터 포털 로그인 데이터 활용 신청 개인 API 인증키 받기 - 추가실습 진행
import requests

# 1. 주소와 내 인증키 설정
url = '공공데이터 포털 공개 데이터 주소' # 주소
my_key = '인증키를 여기에'

# 2. 요청할 조건들 (파라미터)
params = {
    'serviceKey': my_key,
    'returnType' : 'json',    # XML보다 읽기 쉬운 JSON 형식을 추천해요!
    'numOfRows' : '100',
    'pageNo' : '1',
    'year' : '2020',
    'itemCode' : 'PM10'
}

# 3. 데이터 가져오기
response = requests.get(url, params=params)

# 4. 결과 출력
if response.status_code == 200:
    print(response.json()) # 데이터가 쫙 출력됩니다!
else:
    print("오류 발생:", response.status_code)
```
