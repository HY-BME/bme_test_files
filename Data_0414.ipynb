```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 1. 데이터 로드
df = sns.load_dataset("anscombe")

# 2. 통계치 비교 출력
print("=== Anscombe's Quartet Summary Statistics ===")
stats = df.groupby('dataset').agg(['mean', 'std', 'var'])
print(stats)

# 상관계수 계산을 위한 루프
for dataset in ['I', 'II', 'III', 'IV']:
    curr_df = df[df['dataset'] == dataset]
    corr = curr_df['x'].corr(curr_df['y'])
    print(f"\nDataset {dataset} Correlation: {corr:.3f}")

# 3. 시각화 결과 비교
# col_wrap=2를 통해 2x2 격자로 배치합니다.
grid = sns.lmplot(x="x", y="y", col="dataset", hue="dataset", data=df,
                  col_wrap=2, ci=None, palette="muted", height=4,
                  scatter_kws={"s": 50, "alpha": 1},
                  line_kws={"lw": 2})

# 그래프 타이틀 및 레이아웃 조정
grid.set_axis_labels("X Value", "Y Value")
plt.subplots_adjust(top=0.9)
grid.fig.suptitle("Anscombe's Quartet: Same Stats, Different Shapes", fontsize=16)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터 로드
df = pd.read_csv('/content/sample_data/avocado.csv')

# 비교: 타입별 평균 가격 계산 및 시각화
plt.figure(figsize=(8, 5))
sns.barplot(x='type', y='AveragePrice', data=df, palette='viridis')
plt.title('Comparison: Average Price by Type')
plt.show()

print(df.groupby('type')['AveragePrice'].mean())
```

```python
# 구성: 전체 가방 중 사이즈별 총합 계산
bag_sizes = df[['Small Bags', 'Large Bags', 'XLarge Bags']].sum()

# 파이 차트로 시각화
plt.figure(figsize=(7, 7))
plt.pie(bag_sizes, labels=bag_sizes.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99'])
plt.title('Composition: Total Bags by Size')
plt.show()
```

```python
# 분포: 평균 가격의 히스토그램과 커널 밀도 추정(KDE)
plt.figure(figsize=(10, 6))
sns.histplot(df['AveragePrice'], kde=True, color='skyblue')
plt.title('Distribution: Density of Average Price')
plt.xlabel('Average Price')
plt.ylabel('Frequency')
plt.show()
```

```python
# 관계: 판매량과 가격 사이의 산점도 (Scatter Plot)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Total Volume', y='AveragePrice', data=df, alpha=0.5, color='orange')

# x축이 너무 크므로 로그 스케일 적용 (더 명확한 관계 파악을 위해)
plt.xscale('log')
plt.title('Relationship: Total Volume vs Average Price (Log Scale)')
plt.show()
```

```python
# seaborn을 이용하여 titanic 데이터에서 age와 fare의 관계를 나타내는 산점도를 그리고, hue="sex"를 추가하세요.

import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')

plt.figure(figsize=(10, 5))
sns.scatterplot(x='age', y='fare', data=titanic, hue='sex')

plt.title('Age vs Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True)
plt.show()
```

```python
# seaborn을 이용하여 titanic 데이터에서 상관 관계 히트맵을 그리세요.

import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset('titanic')

# numeric_only=True를 추가하여 수치형 컬럼만 계산하도록 수정
plt.figure(figsize=(10, 8))
sns.heatmap(titanic.corr(numeric_only=True), annot=True, cmap='coolwarm')

plt.title('Titanic Dataset Correlation Heatmap')
plt.show()
```
