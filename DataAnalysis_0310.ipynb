```python
import pandas as pd
```

```python
# 'tips.csv' 파일을 읽어 'tips' 변수에 저장합니다.
# 파일이 Colab 환경에 업로드되어 있어야 합니다.
tips = pd.read_csv('/content/sample_data/tips.csv')

# 첫 5행을 출력하여 데이터가 잘 로드되었는지 확인합니다.
print(tips.head(10))
```

```python
print("tips DataFrame의 컬럼 정보:\n")
tips.info()

print("\ntips DataFrame의 통계 요약:\n")
display(tips.describe())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

#-- pyplot : 그래프 사용을 위한 파이썬 라이브러리
plt.figure(figsize=(12, 7)) #-- figsize(width, height) --#
sns.violinplot(x='day', y='total_bill', hue='size', data=tips, palette='viridis', inner='quartile')
plt.title('Total Bill Distribution by Day and Table Size (Violin Plot)')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.grid(axis='y', linestyle='--')
plt.legend(title='Table Size', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
# 'day'를 마커의 모양(style)으로 구분하고, 'size'는 마커의 크기로 유지합니다.
sns.scatterplot(x='total_bill', y='tip', data=tips, style='day', size='size', alpha=0.7)

# 회귀선을 추가하여 전반적인 경향을 표시합니다.
#sns.regplot(x='total_bill', y='tip', data=tips, scatter=False, color='gray', line_kws={'linestyle':'--'})

plt.title('Relationship between Total Bill and Tips)')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.grid(axis='both', linestyle='--', alpha=0.7)
plt.legend(title='Day / Table Size', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
# 'day'를 x축으로, 'tip'을 y축으로 하여 팁의 분포를 시각화하고,
# 'time' (Dinner/Lunch)을 hue로 사용하여 각 요일 내 시간대별로 구분합니다.
sns.violinplot(x='day', y='tip', hue='time', data=tips, palette='viridis', inner='quartile')
plt.title('Distribution of Tips by Day and Time')
plt.xlabel('Day of the Week')
plt.ylabel('Tip ($)')
plt.grid(axis='y', linestyle='--')
plt.legend(title='Time', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
# 'day'를 x축으로, 'tip'을 y축으로 하여 팁의 분포를 시각화하고,
# 'time' (Dinner/Lunch)을 hue로 사용하여 각 요일 내 시간대별로 구분하는 상자 그림을 그립니다.
sns.boxplot(x='day', y='tip', hue='time', data=tips, palette='pastel')
plt.title('Distribution of Tips by Day and Time (Box Plot)')
plt.xlabel('Day of the Week')
plt.ylabel('Tip ($)')
plt.grid(axis='y', linestyle='--')
plt.legend(title='Time', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 1. `pd.read_csv()` 함수를 사용하여 'https://raw.githubusercontent.com/plotly/datasets/master/mtcars.csv' URL에서 데이터를 로드하고 `mtcars_df` 변수에 저장합니다.
mtcars_df = pd.read_csv('https://raw.githubusercontent.com/plotly/datasets/master/mtcars.csv')

# 2. `mtcars_df` DataFrame의 첫 5행을 출력하여 데이터가 올바르게 로드되었는지 확인합니다.
print("mtcars DataFrame의 첫 5행:\n", mtcars_df.head())
```

```python
mtcars_df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 'manufacturer' 컬럼은 object 타입이므로 상관관계 분석에서 제외
correlation_matrix = mtcars_df.drop(columns=['manufacturer']).corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of mtcars Dataset Variables')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 7))
sns.boxplot(x='manufacturer', y='mpg', data=mtcars_df, palette='viridis')
plt.title('MPG Distribution by Manufacturer')
plt.xlabel('Manufacturer')
plt.ylabel('Miles Per Gallon (MPG)')
plt.xticks(rotation=45, ha='right') # 제조사 이름이 겹치지 않도록 회전
plt.grid(axis='y', linestyle='--')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.scatterplot(x='drat', y='mpg', data=mtcars_df, alpha=0.8, s=100)
plt.title('MPG vs. Rear Axle Ratio (drat)')
plt.xlabel('Rear Axle Ratio (drat)')
plt.ylabel('Miles Per Gallon (MPG)')
plt.grid(True, linestyle='--')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.scatterplot(x='drat', y='mpg', hue='cyl', data=mtcars_df, alpha=0.8, s=100, palette='viridis')
plt.title('MPG vs. Rear Axle Ratio (drat) by Cylinder Count')
plt.xlabel('Rear Axle Ratio (drat)')
plt.ylabel('Miles Per Gallon (MPG)')
plt.grid(True, linestyle='--')
plt.legend(title='Cylinders')
plt.tight_layout()
plt.show()
```
